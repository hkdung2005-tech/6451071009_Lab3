## Câu a

```python
print("Hồng Kim Dung - 6451071009")
from sklearn.datasets import load_iris

iris = load_iris()
print("Đã đọc bộ dữ liệu Iris:", type(iris))
```

## Câu b (1)

```python
print("Hồng Kim Dung - 6451071009")
print("Tên các thuộc tính:", iris.feature_names)
print("Tên các lớp:", [str(t) for t in iris.target_names])
```

## Câu b (2)

```python
print("Hồng Kim Dung - 6451071009")
print("data.shape  :", iris.data.shape)
print("target.shape:", iris.target.shape)
print("Số mẫu:", iris.data.shape[0], "| Số thuộc tính:", iris.data.shape[1])
```

## Câu b (3)

```python
print("Hồng Kim Dung - 6451071009")
for i in range(5):
    print(iris.data[i], "->", iris.target[i])
```

## Câu b (4)

```python
print("Hồng Kim Dung - 6451071009")
import pandas as pd

df = pd.DataFrame(iris.data, columns=iris.feature_names)
print(df.describe())
```

## Câu c

```python
print("Hồng Kim Dung - 6451071009")
X = iris.data
Y = iris.target
X = X / X.max(axis=0)   # chia mỗi cột cho giá trị lớn nhất của cột
print("Min:", X.min(axis=0))
print("Max:", X.max(axis=0))
print(X[:5])
```

## Câu d

```python
print("Hồng Kim Dung - 6451071009")
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
print("X_train:", X_train.shape, "| X_test:", X_test.shape)
print("y_train:", y_train.shape, "| y_test:", y_test.shape)
```

## Câu e - Decision Tree

```python
print("Hồng Kim Dung - 6451071009")
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

def danh_gia(ten, y_pred):
    print("Accuracy:", accuracy_score(y_test, y_pred))
    print(classification_report(y_test, y_pred, target_names=iris.target_names))
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=iris.target_names, yticklabels=iris.target_names)
    plt.xlabel("Dự đoán"); plt.ylabel("Thực tế"); plt.title("Confusion matrix - " + ten)
    plt.show()

dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)
danh_gia("Decision Tree", y_pred_dt)
```

## Câu f - Naive Bayes

```python
print("Hồng Kim Dung - 6451071009")
from sklearn.naive_bayes import GaussianNB

nb = GaussianNB()
nb.fit(X_train, y_train)
y_pred_nb = nb.predict(X_test)
danh_gia("Naive Bayes", y_pred_nb)
```

## Câu g - KNN

```python
print("Hồng Kim Dung - 6451071009")
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)
danh_gia("KNN", y_pred_knn)
```

## Câu h - Neural Network (Keras)

```python
print("Hồng Kim Dung - 6451071009")
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

nn = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(10, activation="relu"),
    Dense(20, activation="relu"),
    Dense(3, activation="softmax"),
])
nn.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
nn.summary()

nn.fit(X_train, y_train, epochs=100, batch_size=1, verbose=1)

y_prob_nn = nn.predict(X_test)
y_pred_nn = np.argmax(y_prob_nn, axis=1)
danh_gia("Neural Network", y_pred_nn)
```

## Câu i - Dự đoán 3 mẫu mới

```python
print("Hồng Kim Dung - 6451071009")
X_new = np.array([[6.2, 2.9, 4.3, 1.3],
                  [5.1, 3.5, 1.4, 0.2],
                  [7.3, 2.8, 6.4, 2.1]])
X_new_scaled = X_new / iris.data.max(axis=0)   # chuẩn hóa cùng thang đo với dữ liệu train

ten_lop = iris.target_names
print("Decision Tree:", [str(ten_lop[k]) for k in dt.predict(X_new_scaled)])
print("Naive Bayes  :", [str(ten_lop[k]) for k in nb.predict(X_new_scaled)])
print("KNN          :", [str(ten_lop[k]) for k in knn.predict(X_new_scaled)])
print("Neural Net   :", [str(ten_lop[k]) for k in np.argmax(nn.predict(X_new_scaled), axis=1)])
```
